# Tuplas

### Una tupla es como una lista, pero no se puede modificar una vez creada (es inmutable). Se escribe entre paréntesis ( ). Se usa para datos que no deben cambiar, como coordenadas o una fecha.

```
    punto = (10, 20)          # coordenadas x, y
    print(punto[0])           # 10 (se accede igual que en listas)
    punto[0] = 5  -> ERROR: las tuplas no se pueden modificar
```

````python
punto = (10, 20) # coordenadas x, y
print(f'La Tupla es: {punto}')
print(punto[1])  # 10 (se accede igual que en listas)

# ERROR: las tuplas no se pueden modificar
# punto[0] = 5  -> 

# Se puede crear sin paréntesis
tupla = 1, 2, 3
print(type(tupla)) #<class 'tuple'>
print(tupla)       #(1, 2, 3)
````

````python
# Las tuplas también pueden ser anidadas
tupla = 1, 2, ('a', 'b', 'c', (1, 2, 3)), 3
print(tupla)       #(1, 2, ('a', 'b'), 3)
print(tupla[2][3][1]) #a
````

````python
# Y también es posible convertir una lista en tupla haciendo uso de al función tuple().
lista = [1, 2, 3]
tupla = tuple(lista)
print(lista)
print(type(tupla)) #<class 'tuple'>
print(tupla)       #(1, 2, 3)
````

````python
# Iterar una Tupla
tupla = [1, 2, 3]
for t in tupla:
    print(t) #1, 2, 3
````

````python
numeros = (1, 2, 3)
x, y, z = numeros
print(x) #1
print(y) #2
print(z) #3
````

````python
# Crear una Tupla con 1 solo elemento
tupla = (2,)
print(type(tupla)) #<class 'tuple'>
print(tupla)
````

## Métodos de Tuplas

### count(<obj>)
#### El método count() cuenta el número de veces que el objeto pasado como parámetro se ha encontrado en la lista.

````python
l = (1, 1, 1, 3, 5)
print(l.count(1)) #3
````

### index(<obj>[,index])
#### El método index() busca el objeto que se le pasa como parámetro y devuelve el índice en el que se ha encontrado

````python
l = (7, 7, 7, 3, 5)
print(l.index(5)) #4
````

#### El método index() también acepta un segundo parámetro opcional, que indica a partir de que índice empezar a buscar el objeto.

````python
l = (7, 7, 7, 3, 5)
print(l.index(7, 2, 5)) #2
````
