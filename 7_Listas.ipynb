# Listas en Python

## La lista es la colección más versátil de Python. Es una secuencia ordenada (los elementos tienen una posición fija) y mutable (se puede agregar, eliminar o cambiar elementos). Se crea con corchetes [ ] y los elementos se separan por comas. Una lista puede contener cualquier tipo de dato, incluso mezclar tipos distintos:



```
calificaciones = [7, 9, 6, 10, 8]
materias = ["Matemáticas", "Español", "Programación"]
mixta = [1, "hola", True, 3.14]     # válido, aunque poco usual
vacia = []                           # lista vacía: la irás llenando con append()
```



````python
# Acceder por índice
# Cada elemento tiene una posición llamada índice. El primer elemento es el índice 0 (no 1).
# Los índices negativos cuentan desde el final: -1 es el último elemento, -2 el penúltimo, etc.:

frutas = ["manzana", "pera", "uva", "naranja"]

print(frutas[0])    # manzana (primero)
print(frutas[2])    # uva (tercero)
print(frutas[-1])   # naranja (último)
print(frutas[-2])   # uva (penúltimo)
print(len(frutas))  # 4 (cantidad de elementos)
print(frutas)

'''IndexError. Si intentas acceder a un índice que no existe (por ejemplo frutas[10]
en una lista de 4 elementos), Python lanza IndexError: list index out of range.
Siempre verifica que el índice esté entre 0 y len(lista) - 1.'''
print(frutas[10])
````

## Modificar, agregar y eliminar

```
frutas = ["manzana", "pera", "uva"]

frutas[1] = "plátano"      # cambia el elemento en posición 1
frutas.append("naranja")   # agrega al final
frutas.insert(0, "kiwi")   # inserta "kiwi" en la posición 0
frutas.remove("uva")       # elimina la primera ocurrencia de "uva"
ultimo = frutas.pop()      # elimina y devuelve el último elemento
print(frutas)
print("Eliminado:", ultimo)
```



````python
frutas = ["uva", "uva", 'perro', "uva"]
print('Lista Original:')
print(frutas)

frutas[1] = "plátano"      # cambia el elemento en posición 1
print('\nCambiar el elemento de una Posición 1:')
print(frutas)

frutas.append('Naranja')   # agrega al final
print('\nAgregar un Elemento al Final de la Lista')
print(frutas)

frutas.insert(0, "kiwi")   # inserta "kiwi" en la posición 0
print('\nInsertar un Elemento en una posición determinada 0')
print(frutas)

frutas.remove("uva")       # elimina la primera ocurrencia de "uva"
print('\nElimina la Primera Ocurrencia de un Elemento')
print(frutas)

ultimo = frutas.pop()      # elimina y devuelve el último elemento
print('\nElimina el Último Elemento de la Lista')
print(frutas)
print('Eliminado:', ultimo)

frutas.append(ultimo)
print('\nVolvemos a Agregar al Final el elemento borrado con POP')
print(frutas)
````
