```python
# Usando Split
texto = "hoy comi tacos al pastor"
tokens = texto.split()
print(tokens)
print("Numero de tokens:", len(tokens))
```

```python
# Esto pasa con signos pegados al texto o palabras.
texto = "hola, ¿como estas?"
print(texto.split())
```

```python
# Opción robusta: NLTK

import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
from nltk.tokenize import word_tokenize


texto = "¡hola!, ¿como estas? todo bien."
tokens = word_tokenize(texto, language="spanish")
print(tokens)
```

```python
# Tokenizar por oraciones.
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
from nltk.tokenize import sent_tokenize

texto = "La pelicula estuvo bien. Los efectos no tanto. Aun asi la recomiendo."
oraciones = sent_tokenize(texto, language="spanish")
for i, o in enumerate(oraciones, 1):
    print(i, "->", o)

```
