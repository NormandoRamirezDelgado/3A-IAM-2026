# Diccionarios (clave–valor)
## Un diccionario guarda pares clave: valor, como un diccionario real relaciona una palabra con su definición. Se escribe entre llaves { }. Accedes por la clave, no por posición.
### Ejemplo:

``` 
    alumno = {
        "nombre"    : "Ana",
        "edad"      : 16,
        "promedio"  : 9.2
    }
```

````python
alumno = {
    "nombre"    : "Ana", 
    "edad"      : 16, 
    "promedio"  : 9.2
}

print(alumno)
print(alumno["nombre"])       # Ana (acceso por la clave)
alumno["edad"] = 17           # modificar un valor
alumno["grupo"] = "3°A-IA"        # agregar un nuevo par clave-valor
print(alumno)
````

### Verificar si una clave existe y acceso seguro con .get()

#### Si intentas acceder a una clave que no existe, Python lanza un KeyError. Dos formas de evitarlo:

````python
alumno = {
    "nombre"    : "Ana", 
    "edad"      : 16
}

# 1) Verificar con "in" antes de acceder
if "nombre" in alumno:
    print(alumno["nombre"])
else:
    print("No hay promedio registrado")

# 2) .get(clave, valor_por_defecto): devuelve el valor o el default si no existe
promedio = alumno.get("edad", 0)   # no hay error aunque la clave no exista
if promedio == 0:
    print('No Existe')
else:
    print(promedio)                         # 0
````

````python
# Recorrrer un diccionario con FOR
alumno = {
    'nombre': 'Ana', 
    'edad': 16, 
    'promedio': 9.2
}

for clave, valor in alumno.items():
    print(f"  {clave}: {valor}")
````

````python
# Lista de diccionarios
contactos = [
    {
        "nombre": "Ana",    
        "telefono": "555-1001", 
        "email": "ana@mail.com"
    },
    {
        "nombre": "Luis",   
        "telefono": "555-2002", 
        "email": "luis@mail.com"
    },
    {"nombre": "María",  "telefono": "555-3003", "email": "maria@mail.com"},
]

# Mostrar todos los contactos
for i, contacto in enumerate(contactos, start=1):
    print(f"{i}. {contacto['nombre']} — Tel: {contacto['telefono']}")

# Buscar un contacto por nombre
buscar = "Luis"
for contacto in contactos:
    if contacto["nombre"] == buscar:
        print(f"Encontrado: {contacto['email']}")
````

````python
# Eliminar clave y valor de un Diccionario
alumno = {
    'nombre': 'Ana', 
    'edad': 16, 
    'promedio': 9.2,
    'grupo' : '3A-IAM'
}
print(alumno)
del alumno["grupo"]     # elimina la clave "grupo"
print(alumno)

edad = alumno.pop("edad")   # elimina y devuelve el valor
print(f'Clave Eliminada: {edad}')
print(alumno)

#print(alumno)     # {'nombre': 'Ana'}
````

````python
# Métodos Principales en diccionarios
alumno = {"nombre": "Ana", "edad": 16, "grupo": "3°A"}

print(alumno.keys())     # dict_keys(['nombre', 'edad', 'grupo'])
print(alumno.values())   # dict_values(['Ana', 16, '3°A'])
print(alumno.items())    # dict_items([('nombre', 'Ana'), ('edad', 16), ('grupo', '3°A')])
print(len(alumno))       # 3
````

````python
# Recorrer Diccionario accediendo a solamente valores o claves
alumno = {"nombre": "Ana", "edad": 16, "grupo": "3°A"}

# Solo claves
for clave in alumno:
    print(clave)

# Solo valores
for valor in alumno.values():
    print(valor)
````

````python
# Método .UPDATE Agrega las claves de otro diccionario; si la clave ya existe, sobreescribe el valor.
alumno = {
    "nombre": "Ana", 
    "edad": 16, 
}

alumno.update(
    {
        'edad'  : 17,
        'grupo' : '3A-IAM'
    }
)
print(alumno)
````

````python
# Método .CLEAR() Elimina todos los elementos
alumno = {
    "nombre": "Ana", 
    "edad": 16, 
}
alumno.clear()
print(alumno)
````
