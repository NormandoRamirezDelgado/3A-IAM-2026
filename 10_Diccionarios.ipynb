# Diccionarios (clave–valor)
## Un diccionario guarda pares clave: valor, como un diccionario real relaciona una palabra con su definición. Se escribe entre llaves { }. Accedes por la clave, no por posición.
### Ejemplo:

### alumno = {
###    "nombre": "Ana",
###    "edad": 16,
###    "promedio": 9.2
### }

```python
alumno = {
    "nombre"    : "Ana", 
    "edad"      : 16, 
    "promedio"  : 9.2
}

print(alumno)
print(alumno["nombre"])       # Ana (acceso por la clave)
alumno["edad"] = 17           # modificar un valor
alumno["grupo"] = "3°A-IA"        # agregar un nuevo par clave-valor
print(alumno)
```

### Verificar si una clave existe y acceso seguro con .get()

#### Si intentas acceder a una clave que no existe, Python lanza un KeyError. Dos formas de evitarlo:

```python
alumno = {
    "nombre"    : "Ana", 
    "edad"      : 16
}

# 1) Verificar con "in" antes de acceder
if "nombre" in alumno:
    print(alumno["nombre"])
else:
    print("No hay promedio registrado")

# 2) .get(clave, valor_por_defecto): devuelve el valor o el default si no existe
promedio = alumno.get("edad", 0)   # no hay error aunque la clave no exista
if promedio == 0:
    print('No Existe')
else:
    print(promedio)                         # 0
```
