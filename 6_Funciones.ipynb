# Funciones en Python

### Piensa en una receta con nombre 📖: la escribes una sola vez y la usas cuantas veces quieras. Una función es un bloque de código con nombre que se define una vez con def y se llama escribiendo su nombre con paréntesis. Puede recibir datos (parámetros) y devolver un resultado (return).
Así se escribe una función: la palabra def, el nombre, los paréntesis (con los parámetros adentro, si los hay), dos puntos : y el cuerpo indentado. Lo indentado es lo que se ejecuta cada vez que la llamas.

```python
def nombreFuncion():
  # Código de la función
  print('Bienvenido a las Funciones!!!!')

def mensaje():
  print('Esta función solo es una prueba')

# Llamar la Función
nombreFuncion()
mensaje()

print(mensaje())
```

```python
# Retorno de valor de una función
def suma():
  uno = 10
  dos = 15
  return uno + dos

print(f'el Resultado de la Suma es: {suma()}')
print(suma() * 10)
retorno = suma()
print(retorno)
```

```python
# Funciones con Parámetros.
def saludar(nombre):            # "nombre" es el parámetro
    print(f"¡Hola, {nombre}! Bienvenido.")

saludar('Ana')                  # "Ana" es el argumento
saludar('Luis')                 # se reutiliza con otro argumento
saludar('Profe')                # misma función, resultado diferente

argumento = input('Nombre de la Persona:')
saludar(argumento)
```

```python
# El Retorno de una función como Argumento
def suma(uno, dos):
   return uno + dos

def resta(uno, dos):
  return uno - dos

uno = int(input('Valor Uno:'))
dos = int(input('Valor Dos:'))
print(f'Suma: {suma(uno, dos)}')
print(f'Resta: {resta(uno, suma(uno, dos))}')
```

```python
# Parámetros con valor por Defecto
def saludar_con_titulo(nombre, titulo="Alumno"):
    print(f"Bienvenido, {titulo} {nombre}")

saludar_con_titulo("López", "Profesor")    # sobreescribe el default: Profesor López
saludar_con_titulo("García")               # usa el default: Alumno García
```
