```python
# La lista de NLTK 
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords

vacias = stopwords.words("spanish")

print("Cuántas hay:", len(vacias))
print("Primeras 10:", vacias[:10])
print("¿Está 'no'?", "no" in vacias)
```

```python
# Filtra un texto
texto = "el gato duerme en la casa de mi abuela"
tokens = texto.split()
importantes = [t for t in tokens if t not in vacias]
print("Original:      ", len(tokens), tokens)
print("Sin stop words:", len(importantes), importantes)
```

```python
# Cuando quitar stop words destruye el mensaje
a = "me gusto mucho la pelicula"
b = "no me gusto nada la pelicula"
print([t for t in a.split() if t not in vacias])
print([t for t in b.split() if t not in vacias])
```

```python
# Haz tu propia lista
mis_vacias = set(stopwords.words("spanish"))
for palabra in ["no", "nada", "mucho", "muy"]:
    mis_vacias.discard(palabra)
mis_vacias.add("rt")

print("Quedan", len(mis_vacias), "stop words")
print([t for t in a.split() if t not in mis_vacias])
print([t for t in b.split() if t not in mis_vacias])
```

```python
# Tu pipeline completo: procesar()
import re

def limpiar(texto):
    texto = texto.lower()
    texto = re.sub(r"https?://\S+", " ", texto)
    texto = re.sub(r"@\w+", " ", texto)
    texto = re.sub(r"[^a-záéíóúñ ]", " ", texto)
    texto = re.sub(r"(\w)\1{2,}", r"\1", texto)
    texto = re.sub(r"\s+", " ", texto).strip()
    return texto


def procesar(texto):
    limpio = limpiar(texto)
    tokens = limpio.split()
    utiles = [t for t in tokens if t not in mis_vacias]
    return utiles

comentario = "RT @cafe: Los tacos NO estaban ricos!!! y la fila muuuy larga 😡 https://t.co/x1"
print("1 crudo:  ", comentario)
print("2 limpio: ", limpiar(comentario))
print("3 tokens: ", limpiar(comentario).split())
print("4 útiles: ", procesar(comentario))
```

```python
# El pipeline sobre tu corpus
mi_corpus = [
    "Los tacos de la cafetería están muy ricos",
    "La torta estaba fría y el pan duro",
    "Muy rica la torta de milanesa",
    "Los tacos estaban fríos",
    "El agua de horchata está muy rica",
    "La fila es muy larga y tardan mucho",
    "Rica comida pero muy cara",
    "El servicio es rápido y amable",
]

crudo, limpios, utiles = [], [], []
for documento in mi_corpus:
    crudo = crudo + documento.split()
    limpios = limpios + limpiar(documento).split()
    utiles = utiles + procesar(documento)

print("Paso            Tokens  Vocabulario")
print(f"Crudo           {len(crudo):>6}  {len(set(crudo)):>11}")
print(f"Limpio          {len(limpios):>6}  {len(set(limpios)):>11}")
print(f"Sin stop words  {len(utiles):>6}  {len(set(utiles)):>11}")
```

```python
# Cuáles son las palabras más frecuentes
conteo = {}
for token in utiles:
    conteo[token] = conteo.get(token, 0) + 1

repetidas = [p for p in conteo if conteo[p] >= 2]
print("Aparecen 2 veces o más:", repetidas)
```
