```python
# Espresión Regular "REGEX"
import re

texto = "hola    mundo   feliz"
un_espacio = re.sub(r"\s+", " ", texto)   # varios espacios -> uno solo
print(un_espacio)
```

```python
# Quitar todo lo que no sea letra ni espacio
import re

texto = "¡Hola!! ¿Qué tal el día 100%? 😃 #genial"
texto = texto.lower()
limpio = re.sub(r"[^a-záéíóúñ ]", "", texto)   # deja solo letras y espacios
limpio = re.sub(r"\s+", " ", limpio).strip()   # arregla espacios
print(limpio)
```

```python
# Aplanar Acéntos o eliminarlos
import unicodedata

def quitar_acentos(texto):
    # separa cada letra de su acento y borra los acentos
    nfd = unicodedata.normalize("NFD", texto)
    return "".join(c for c in nfd if unicodedata.category(c) != "Mn")

print(quitar_acentos("canción de después"))
```

```python
import unicodedata

def quitar_acentos_cuidando_ñ(texto):
    print(f'Texto Original: {texto}')
    texto = texto.replace("ñ", "@@").replace("Ñ", "@@")   # la escondemos
    print(f'Ocultamos ñ: {texto}')
    nfd = unicodedata.normalize("NFD", texto)
    print(f'Eliminamos acéntos y ñ: {nfd}')
    sin = "".join(c for c in nfd if unicodedata.category(c) != "Mn")
    print(f'Tíldes y marcas especiales: {sin}')
    final = sin.replace("@@", "ñ")
    print(f'Regresamos las ñ: {final}')
    return final      # la devolvemos

print(quitar_acentos_cuidando_ñ("el niño cumplió un año en España"))
```
