```python
# Ejemplo sencillo de Pipeline
texto = "El PERRO corre. El perro ladra y grita!"

# Paso 2: limpieza basica -> minusculas
texto = texto.lower()
print("Limpio:", texto)

# Paso 3: tokenizar -> partir en palabras
tokens = texto.split()
print("Tokens:", tokens)

# Paso 4 (simplificado): quitar dos palabras muy comunes
comunes = ["el", "la", 'y']
importantes = [t for t in tokens if t not in comunes]
print("Sin palabras comunes:", importantes)
```
