```python
import string

texto = "¡hola, mundo! ¿todo bien?"

resultado = []
for c in texto:
    if c not in string.punctuation + '¿¡':
        resultado.append(c)
sin_signos = ''.join(resultado)
print(sin_signos)
```

```python
# Medir el efecto de limpiar
import string
signos = string.punctuation + "¡¿"


texto = "Perro perra PERRO perro, perro."

antes = texto.split()
print("Antes:  ", len(set(antes)), "palabras distintas ->", sorted(set(antes)))

limpio = texto.lower()
limpio = "".join([c for c in limpio if c not in signos])
print(limpio)
despues = limpio.split()
print(despues)
#Obtengo cuantos elementos distintos hay en la lista
print(set(despues))
print("Después:", len(set(despues)), "palabra distinta   ->", sorted(set(despues)))
```

```python
# Un comentario real de redes, línea por línea
import re
texto = "@ana JAJAJA me encantooo!!! la lllave 😍 mira https://t.co/abc/asdfasdf/sdf.html #cine"

texto = texto.lower()
print("1 minúsculas:    ", texto)

texto = re.sub(r"https?://\S+", " ", texto)
print("2 sin enlaces:   ", texto)

texto = re.sub(r"@\w+", " ", texto)
print("3 sin menciones: ", texto)

texto = re.sub(r"[^a-záéíóúñ ]", " ", texto)
print("4 solo letras:   ", texto)

texto = re.sub(r"(\w)\1{2,}", r"\1", texto)
print("5 sin alargadas: ", texto)

texto = re.sub(r"\s+", " ", texto).strip()
print("6 un solo espacio:", texto)
```

```python
# Empaca todo en tu función limpiar()
import re

def limpiar(texto):
    texto = texto.lower()
    texto = re.sub(r"https?://\S+", " ", texto)
    texto = re.sub(r"@\w+", " ", texto)
    texto = re.sub(r"[^a-záéíóúñ ]", " ", texto)
    texto = re.sub(r"(\w)\1{2,}", r"\1", texto)
    texto = re.sub(r"\s+", " ", texto).strip()
    return texto

pruebas = [
    "  Me ENCANTA el PLN!!!  ",
    "Los tacos están RIQUÍSIMOS 🌮🌮 10/10",
    "nooooo, la torta estaba fría... @cafeteria",
]
for prueba in pruebas:
    print(repr(limpiar(prueba)))
```

```python
# Opcional: quitar acentos
import unicodedata

def quitar_acentos(texto):
    separado = unicodedata.normalize("NFD", texto)
    return "".join([c for c in separado if unicodedata.category(c) != "Mn"])

print(quitar_acentos("canción de después"))
print(quitar_acentos("el niño cumplió un año"))
```
