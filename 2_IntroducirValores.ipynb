# Input

## Introduce valores desde teclado

## Sintáxis
### variable = input('Mensaje')

#### Todo lo introducido por el teclado en un input siempre será de tipo String o Cadena


```python
# Pedir valores
nombre = input('Cuál es tu Nombre? ')
semestre = input('Cuál tu Semestre? ')
especialidad = input('Cuál es tu Especialidad? ')
```

```python
# Introducir valores numéricos
anioNacimiento = int(input('En que año naciste? '))
edad = 2026 - anioNacimiento
print(f'tu edad es: {edad}')

# Introducir números decimales
venta = float(input('Venta Realizada: '))
descuento = venta * 0.16
total = venta - descuento
print(f'La venta es de: ${total}')

# Convertir un número a cadena
numero = 100
cadena = str(numero)
resta = cadena - 10
print(f'La resta es: {resta}')
```
