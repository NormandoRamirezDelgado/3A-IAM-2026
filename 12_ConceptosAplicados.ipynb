```python
documento = "Hoy comi tacos. Hoy fue un buen dia"
tokens = documento.split() # partimos en palabras
print("Tokens:", tokens)
print("Numero de tokens:", len(tokens))

vocabulario = set(tokens) # palabras distintas
print("Vocabulario:", vocabulario)
print("Tamano del vocabulario:", len(vocabulario))
```

```python
# Normalizar un Texto
print(len(set("Hoy hoy HOY".split())))          # sin normalizar
print(len(set("Hoy hoy HOY".lower().split())))  # normalizando primero
```

```python
corpus = [
    "hoy comi tacos",
    "hoy comi pizza",
    "los tacos estaban ricos"
]

todos_los_tokens = []
for documento in corpus:
    todos_los_tokens = todos_los_tokens + documento.split()

vocabulario = set(todos_los_tokens)
print('Corpus:', corpus)
print("Documentos:  ", len(corpus))
print('Todos los Tokens:', todos_los_tokens)
print("Tokens:      ", len(todos_los_tokens))
print('Vocabulario:', vocabulario)
print("Vocabulario: ", len(vocabulario))
print("Riqueza:     ", round(len(vocabulario) / len(todos_los_tokens), 2))
```
