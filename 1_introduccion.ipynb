# Variables

### Formas correctas de definir nombres de variables.

- Sin acéntos
- No puede empezar con números, o cualquier otro caracter especial.
- no puede llevar espacios
- siempre en minúsculas
- no abreviar los nombre
- no utilizar números arábigos
- no utilizar literales
- deben tener un nombre que este relacionado con lo que se va a almacenar en ella
