# Cadenas String

### Ya la vimos: es texto entre comillas. El texto tiene longitud (cuántos caracteres tiene), que se mide con len(...). Un carácter es cada letra, número, espacio o símbolo.



```
# Obtenemos la longitud de la cadena
frase = "Hola Mexico"
print(len(frase))

```



````python
frase = 'Hola Mundo!'
print(frase)
# Cuenta e imprime el resultado
print(f'Longitud de Caracteres: {len(frase)}')

# Solo cuenta los caracteres
len(frase)

# Necesito el resultado de la longitud se guarde
total = len(frase)

# Puedo imprimir el valor de la variable que contiene la longitud
print(total)
````

### Ojo con esto, que confunde a todos: len() sobre una cadena cuenta caracteres; sobre una lista cuenta elementos. La misma función, dos significados distintos según lo que le des:

```
frase = "Hola Mexico"
print(len(frase))           # 11 caracteres
print(len(frase.split()))   # 2 palabras

```



````python
frase = "Hola Mexico"
print(len(frase))           # 11 caracteres
print(len(frase.split()))   # 2 palabras
````

### Concatenar Cadenas

#### Es la unión de 2 o mas cadenas mediante el operador de + y generando una sola cadena de salida que se puede imprimir o asignar a otra nueva cadena.



```
cadenaUno = 'Hola'
cadenaDos = 'Mundo!'
nuevaCadena = cadenaUno + cadenaDos
```



````python
# Concateno 3 cadenas
print('Hola' + ' ' + 'Mundo')

cadenaUno = 'Hola'
cadenaDos = 'Mundo'
print(cadenaUno + cadenaDos)
````

### Repetir Cadenas

#### Mediante el uso de un asterisco puedo repetir una cantidad determinada de cadenas que se especifique y se requiere la cadena a repetir y la cantidad de veces que se desea



```
print('ja' * 3)
```



````python
# Repetir una cadena
cadena = 'ja ' * 5
print(cadena)

# O así se puede
print('ja' * 5)
````

### Mayúsculas y Minúsculas

#### Manipula las mayúsculas y minúsculas de una cadena



```
mensaje = "hola MUNDO"
mensaje.upper()        # 'HOLA MUNDO'
mensaje.lower()        # 'hola mundo'
mensaje.capitalize()   # 'Hola mundo'
mensaje.title()        # 'Hola Mundo'
mensaje.swapcase()     # 'HOLA mundo'
```



````python
mensaje = "hola mundo"
print(mensaje.upper())        # 'HOLA MUNDO'
print(mensaje.lower())        # 'hola mundo'
print(mensaje.capitalize())   # 'Hola mundo'
print(mensaje.title())        # 'Hola Mundo'
print(mensaje.swapcase())     # 'HOLA mundo'
print(mensaje)

mensaje = input('escribe algo: '.capitalize()).upper()
print(mensaje)
````

### Limpieza de Espacios en una cadena


```
s = "   texto   "
s.strip()    # 'texto'      (ambos lados)
s.lstrip()   # 'texto   '   (izquierda)
s.rstrip()   # '   texto'   (derecha)
```



````python
# Elimina el espacio en blanco de una cadena
cadena = '   texto   '
print('|' + cadena + '|')
print('|' + cadena.strip() + '|').   # 'texto'   (izquierda y derecha)
print('|' + cadena.lstrip() + '|')   # 'texto   '   (izquierda)
print('|' + cadena.rstrip() + '|')   # '   texto'   (derecha)
````

### Búsqueda de caracteres en una cadena


```
s = "banana"
s.find("na")        # 2   (primera posición; -1 si no existe)
s.index("na")       # 2   (igual, pero lanza ValueError si no existe)
s.count("a")        # 3
s.startswith("ba")  # True
s.endswith("na")    # True
```



````python
cadena = "banana"

print(f'En que posición se encuentra: {cadena.find('na')}')   # 2   (primera posición; -1 si no existe)
print(f'En que posición se encuentra: {cadena.index('na')}')  # 2   (igual, pero lanza ValueError si no existe)
print(f'Cuántas veces se repite: {cadena.count('a')}')        # 3
print(f'La cadena empieza con: {cadena.startswith('ba')}')  # True
print(f'La cadena termina con: {cadena.endswith('na')}')    # True
````

````python
# Reemplazo de una cadena por otra dentro de una cadena completa

print('me gusta Java'.replace('Java', 'Python'))   # 'me gusta Python'
````
