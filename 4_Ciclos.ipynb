# Ciclos en Python

## for i in range(valor):
## &nbsp;&nbsp;&nbsp;instrucció(es)

```python
# For básico
for i in range(10):
  print(i)
```

```python
# For con valor inicial y final
for i in range(1, 10):
  print(i)
```

```python
# For con valor inicial, final e intervalos
for i in range(2, 11, 2):
  print(i)
```

```python
#For con inicio, fin, decremento
for i in range(10, -21, -1):
  print(i)
```

```python
#Ejemplo de Calcular Promedio de 50 Alumnos
suma = 0
for i in range(5):
  calificacion = float(input('Calificación: ')) # Pedir Calificacion
  suma = suma + calificacion # Sumar la Calificación Solicitada
promedio = suma / 5
print(f'El Promedio es: {promedio}')
```

```python
#Ejemplo de Calcular Promedio de 50 Alumnos donde cada alumno lleva 5 materias
sumaPromedios = 0
for i in range(1, 6): # Alumnos
  suma = 0
  print(f'Introduce Calificaciones del Alumno {i}')
  for j in range(1, 3): # Materias
    calificacion = float(input(f'Calificación {j}: ')) # Pedir Calificacion
    suma = suma + calificacion # Sumar la Calificación Solicitada

  promedioAlumno = suma / 2
  sumaPromedios = sumaPromedios + promedioAlumno

promedioGrupal = sumaPromedios / 5
print(f'El Promedio es: {promedioGrupal}')
```
